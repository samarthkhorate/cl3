```python
# Sample input text
text = "hello world hello samarth"

# MAP PHASE
mapped = []
words = text.split()

for word in words:
    mapped.append((word, 1))

print("Mapped Output:", mapped)

# SHUFFLE & SORT
shuffle = {}
for key, value in mapped:
    shuffle.setdefault(key, []).append(value)

print("After Shuffle:", shuffle)

# REDUCE PHASE
reduced = {}
for key, values in shuffle.items():
    reduced[key] = sum(values)

print("Word Count:", reduced)
```
